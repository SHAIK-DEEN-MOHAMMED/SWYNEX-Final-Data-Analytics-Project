```python
import pandas as pd
import numpy as np
df = pd.read_csv('/content/drive/MyDrive/netflix_titles.csv/netflix_titles.csv')
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df[df['rating'].str.contains('min', na=False)]
```

```python
df  = df.drop_duplicates()
df.shape
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
df.info()
```

```python
wrong = df['rating'].str.contains('min', na=False)
df.loc[wrong, 'duration'] = df.loc[wrong, 'rating']
df.loc[wrong, 'rating'] = np.nan
```

```python
df[df['rating'].str.contains('min', na=False)]
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Not Rated')
```

```python
df = df.dropna(subset=['date_added', 'duration'])
df.shape
```

```python
df.isnull().sum()
df.shape
```

```python
df.to_csv('/content/drive/MyDrive/netflix_cleaned.csv', index=False)
```
